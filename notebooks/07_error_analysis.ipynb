### 3. Protocolo de Análisis Cualitativo y Cuantitativo de Errores
Para el análisis de errores exigido en la Sección 13:
1. **Predicciones correctas con alta confianza**: Representan especies con características diagnósticas conspicuas (e.g. plumaje brillante, morfología de alas única).
2. **Predicciones correctas con baja confianza**: Especies con fondos ambiguos o iluminación subóptima donde la red duda entre taxones cercanos.
3. **Predicciones incorrectas con alta confianza**: El modelo confunde especies hermanas (*sibling species*) del mismo género botánico o zoológico que difieren únicamente en proporciones milimétricas o patrones microscópicos no resolubles a $224 \times 224$.
4. **Patrones de fondo complejo y escala**: Individuos camuflados o donde el organismo ocupa menos del 10% del campo visual presentan la tasa de error más elevada.
